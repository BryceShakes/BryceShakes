```python
tsta = []
for name in tst['item_name'][:100]:
    name = re.sub("[^A-Za-z0-9 .]", "" , name.lower()) # replace all punctation (apart from fullstops, we want to keep numbers whole)
    name = re.sub(' +', ' ', name).strip()
    tsta.append(name.split(' '))
```

```python
wnl = WordNetLemmatizer()
wnl.lemmatize("womens")
```

```python
wnl = WordNetLemmatizer()
new = []
for n,i in enumerate(tsta):
    a = [wnl.lemmatize(word) for word in i]
    new.append(a)
```

```python
p = nltk.PorterStemmer()
new2 = []
for n,i in enumerate(tsta):
    a = [p.stem(word) for word in i]
    new2.append(a)
```

```python
for i in range(len(new)):
    print(new[i], '\n' , new2[i], '\n')
```
